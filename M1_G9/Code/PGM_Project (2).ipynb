```python
# %% [markdown]
# # CausalLOB — M1 Proposal + Data Foundation
#
# **Track:** Applied AI
# **Domain:** Limit-order-book market microstructure
# **Principal dataset:** FI-2010, NoAuction, Decimal Precision
# **PGM objective:** learn a temporally ordered probabilistic dependency graph over market-state variables and use a Dynamic Bayesian Network for probabilistic inference over future price movement.
#
# This notebook is the **M1 data foundation only**. It deliberately stops before causal structure learning.
#
# ### M1 research question
# > Which lagged limit-order-book states are stable probabilistic antecedents of short-horizon midpoint price formation, and can a Dynamic Bayesian Network represent those dependencies for calibrated probabilistic inference?
#
# ### PGM map
# - **Representation:** temporal graphical model / DBN over spread, depth, queue imbalance, quote-flow proxy, and return.
# - **Learning:** latent-confounder-aware temporal structure discovery with LPCMCI; later Bayesian/regularized parameter learning.
# - **Inference:** posterior probabilities of future price states; later exact-vs-approximate inference comparison under partial observation.
#
# ### Important causal scope
# The dataset is observational. The project seeks **robust observational lagged temporal causal candidates under explicit assumptions**, not experimental causal truth.
#
# ### Why this notebook is rebuilt
# The preprocessing is intentionally stricter than the original version:
# 1. uses only raw LOB rows 1–40 plus labels 145–149;
# 2. never crosses inferred day/security boundaries;
# 3. calls the 50 blocks **boundary-separated sequences**, not independent replications;
# 4. keeps `event_volatility_20` as a **regime/sensitivity variable**, not a primary causal primitive;
# 5. replaces nested `imbalance_l5` in the primary causal set with **incremental levels 2–5 imbalance**, reducing algebraic overlap with L1 imbalance;
# 6. names OFI correctly as a **depth-normalized quote-change OFI proxy**;
# 7. saves versioned checkpoints to Google Drive so expensive work is reused after runtime disconnects.
#
# Run top-to-bottom once. On later sessions, completed checkpoints load automatically.

# %%
# ======================================================================================
# CELL 1 — INSTALL, IMPORT, DRIVE, CONFIGURATION
# ======================================================================================

import sys, subprocess, importlib.util, warnings, time, json, hashlib, gc, shutil, zipfile
from pathlib import Path

REQUIRED_PACKAGES = {
    "kagglehub": "kagglehub",
    "pyarrow": "pyarrow",
}

for module_name, pip_name in REQUIRED_PACKAGES.items():
    if importlib.util.find_spec(module_name) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pip_name])

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import kagglehub

try:
    from IPython.display import display
except Exception:
    display = print

warnings.filterwarnings("ignore")

try:
    from google.colab import drive
    if not Path("/content/drive/MyDrive").exists():
        drive.mount("/content/drive")
    else:
        print("Google Drive already mounted.")
except Exception:
    pass

T0 = time.time()

# ------------------------------- USER SETTINGS ----------------------------------------
GROUP_NAME = "G9"
FORCE_REBUILD = False

ROOT = Path("/content/drive/MyDrive/CausalLOB/M1_G9")
CHECKPOINT_DIR = ROOT / "checkpoints"
RESULTS_DIR = ROOT / "results"
GITHUB_DIR = ROOT / "github" / f"M1_{GROUP_NAME}"

for p in [ROOT, CHECKPOINT_DIR, RESULTS_DIR, GITHUB_DIR]:
    p.mkdir(parents=True, exist_ok=True)

CHECKPOINT_VERSION = "M1_DATA_v2_2026-09-14"

EXPECTED_ROWS = 394_337
EXPECTED_DAYS = 10
EXPECTED_SEGMENTS_PER_DAY = 5
EXPECTED_SEQUENCES = 50

DATASET_HANDLE = "ulfricirons/fi-2010"

RAW_CKPT = CHECKPOINT_DIR / "01_canonical_raw.npz"
BOUNDARY_CKPT = CHECKPOINT_DIR / "02_boundaries.npz"
FEATURE_PARQUET = CHECKPOINT_DIR / "03_features.parquet"
FEATURE_NPZ = CHECKPOINT_DIR / "03_features.npz"
TEMPORAL_NPZ = CHECKPOINT_DIR / "04_temporal_primary.npz"
ACF_CSV = CHECKPOINT_DIR / "04_acf.csv"
QUALITY_JSON = CHECKPOINT_DIR / "05_quality_report.json"
SOURCE_MANIFEST = CHECKPOINT_DIR / "source_manifest.json"
RUN_MANIFEST = CHECKPOINT_DIR / "RUN_MANIFEST.json"

REQUIRED_NAMES = (
    ["Train_Dst_NoAuction_DecPre_CF_1.txt"]
    + [f"Test_Dst_NoAuction_DecPre_CF_{i}.txt" for i in range(1, 10)]
)
REQUIRED_SET = set(REQUIRED_NAMES)

RAW_FEATURE_NAMES = []
for level in range(1, 11):
    RAW_FEATURE_NAMES += [
        f"ask_price_l{level}",
        f"ask_volume_l{level}",
        f"bid_price_l{level}",
        f"bid_volume_l{level}",
    ]

LABEL_NAMES = [
    "label_h10",
    "label_h20",
    "label_h30",
    "label_h50",
    "label_h100",
]

# Cleaner causal representation for M1.
PRIMARY_CAUSAL_VARS = [
    "log_relative_spread",
    "log_total_depth_l5",
    "imbalance_l1",
    "imbalance_l2_l5",
    "quote_ofi_depth_norm",
    "log_mid_return_1",
]

REGIME_VARS = [
    "event_volatility_20",
]

print("=" * 96)
print("CausalLOB — M1 DATA FOUNDATION")
print("=" * 96)
print("Checkpoint root :", CHECKPOINT_DIR)
print("Results root    :", RESULTS_DIR)
print("Version         :", CHECKPOINT_VERSION)
print("Force rebuild   :", FORCE_REBUILD)
print("\nPrimary causal variables:")
for i, v in enumerate(PRIMARY_CAUSAL_VARS, 1):
    print(f" {i}. {v}")
print("Regime/sensitivity variables:", REGIME_VARS)

# %%
# ======================================================================================
# CELL 2 — HELPERS + CANONICAL FI-2010 ACQUISITION
# Checkpoint: 01_canonical_raw.npz
# ======================================================================================

def sha256_file(path, chunk_size=1 << 20):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            block = f.read(chunk_size)
            if not block:
                break
            h.update(block)
    return h.hexdigest()

def atomic_savez_compressed(path, **kwargs):
    path = Path(path)
    tmp = path.with_name(path.name + ".tmp.npz")
    np.savez_compressed(tmp, **kwargs)
    tmp.replace(path)

def atomic_json(path, obj):
    path = Path(path)
    tmp = path.with_suffix(path.suffix + ".tmp")
    tmp.write_text(json.dumps(obj, indent=2), encoding="utf-8")
    tmp.replace(path)

def validate_version(z):
    if "checkpoint_version" not in z.files:
        return False
    return str(z["checkpoint_version"].item()) == CHECKPOINT_VERSION

def find_required_txt_files(root):
    root = Path(root)
    found = {}
    if not root.exists():
        return found
    for path in root.rglob("*.txt"):
        if path.name in REQUIRED_SET:
            found[path.name] = path
    return found

def extract_required_from_zip(zip_path, output_dir):
    extracted = {}
    try:
        with zipfile.ZipFile(zip_path, "r") as zf:
            members = zf.namelist()
            basename_to_member = {Path(m).name: m for m in members}

            for required_name in REQUIRED_NAMES:
                if required_name in basename_to_member:
                    destination = Path(output_dir) / required_name
                    if not destination.exists():
                        with zf.open(basename_to_member[required_name]) as src, open(destination, "wb") as dst:
                            shutil.copyfileobj(src, dst)
                    extracted[required_name] = destination

            nested = [
                m for m in members
                if m.lower().endswith(".zip")
                and ("decpre" in m.lower() or "noauction" in m.lower())
            ]

            for member in nested:
                nested_path = Path(output_dir) / Path(member).name
                if not nested_path.exists():
                    with zf.open(member) as src, open(nested_path, "wb") as dst:
                        shutil.copyfileobj(src, dst)
                extracted.update(extract_required_from_zip(nested_path, output_dir))

    except zipfile.BadZipFile:
        pass

    return extracted

def locate_or_download_sources():
    local_roots = [
        Path("/content"),
        Path("/content/fi2010_download"),
        Path("/content/fi2010_decpre"),
    ]

    found = {}
    for root in local_roots:
        found.update(find_required_txt_files(root))

    if len(found) == 10:
        return found

    print(f"Only {len(found)}/10 Decimal-Precision source files found locally.")
    print("Downloading FI-2010 Kaggle mirror once...")

    dataset_path = Path(kagglehub.dataset_download(DATASET_HANDLE))
    found.update(find_required_txt_files(dataset_path))

    extract_dir = Path("/content/fi2010_decpre")
    extract_dir.mkdir(parents=True, exist_ok=True)

    if len(found) < 10:
        for zp in dataset_path.rglob("*.zip"):
            found.update(extract_required_from_zip(zp, extract_dir))
            if len(found) == 10:
                break

    if len(found) < 10:
        for zp in Path("/content").rglob("*.zip"):
            found.update(extract_required_from_zip(zp, extract_dir))
            if len(found) == 10:
                break

    found.update(find_required_txt_files(Path("/content")))

    missing = REQUIRED_SET - set(found)
    if missing:
        raise FileNotFoundError(
            "Missing required FI-2010 Decimal-Precision files:\n"
            + "\n".join(sorted(missing))
        )

    return found

def load_fi2010_matrix(path):
    with open(path, "r", encoding="utf-8", errors="ignore") as f:
        first_line = f.readline()
    first_line_count = len(first_line.split())

    flat = np.fromfile(path, dtype=np.float32, sep=" ")
    if flat.size % 149 != 0:
        raise ValueError(f"{path.name}: total values {flat.size:,} not divisible by 149.")

    if first_line_count == 149:
        matrix = flat.reshape(-1, 149).T
    else:
        matrix = flat.reshape(149, -1)

    if matrix.shape[0] != 149:
        raise RuntimeError(f"{path.name}: unexpected matrix shape {matrix.shape}")

    return matrix

def build_canonical_raw_checkpoint():
    found = locate_or_download_sources()

    day_file_map = {1: found["Train_Dst_NoAuction_DecPre_CF_1.txt"]}
    for cf in range(1, 10):
        day_file_map[cf + 1] = found[f"Test_Dst_NoAuction_DecPre_CF_{cf}.txt"]

    X_days, Y_days, days, sample_ids = [], [], [], []
    source_rows = []

    for d in range(1, 11):
        path = day_file_map[d]
        print(f"  Loading day {d}: {path.name}")
        M = load_fi2010_matrix(path)

        # Official FI-2010 layout:
        # rows 1-40 raw LOB; rows 41-144 publisher engineered features; rows 145-149 labels.
        Xd = M[0:40, :].T.astype(np.float32, copy=True)
        Yd = np.rint(M[144:149, :].T).astype(np.int8)

        if Xd.shape[1] != 40:
            raise RuntimeError("Expected 40 raw LOB columns.")
        if not np.isfinite(Xd).all():
            raise RuntimeError(f"Non-finite raw LOB values on day {d}.")
        if not np.isin(Yd, [1, 2, 3]).all():
            raise RuntimeError(f"Unexpected labels on day {d}: {np.unique(Yd)}")

        n = len(Xd)
        X_days.append(Xd)
        Y_days.append(Yd)
        days.append(np.full(n, d, dtype=np.int8))
        sample_ids.append(np.arange(1, n + 1, dtype=np.int32))

        source_rows.append({
            "day": d,
            "filename": path.name,
            "n_observations": n,
            "sha256": sha256_file(path),
        })

        del M
        gc.collect()

    X_raw = np.concatenate(X_days)
    Y = np.concatenate(Y_days)
    day = np.concatenate(days)
    sample_in_day = np.concatenate(sample_ids)

    if X_raw.shape != (EXPECTED_ROWS, 40):
        raise RuntimeError(f"Canonical raw shape mismatch: {X_raw.shape}")
    if Y.shape != (EXPECTED_ROWS, 5):
        raise RuntimeError(f"Canonical label shape mismatch: {Y.shape}")
    if sorted(np.unique(day).tolist()) != list(range(1, 11)):
        raise RuntimeError("Expected exactly days 1..10.")

    atomic_savez_compressed(
        RAW_CKPT,
        checkpoint_version=np.array(CHECKPOINT_VERSION),
        X_raw=X_raw,
        Y=Y,
        day=day,
        sample_in_day=sample_in_day,
        raw_feature_names=np.array(RAW_FEATURE_NAMES),
        label_names=np.array(LABEL_NAMES),
    )

    atomic_json(SOURCE_MANIFEST, {
        "checkpoint_version": CHECKPOINT_VERSION,
        "dataset": "FI-2010",
        "variant": "NoAuction Decimal Precision",
        "kaggle_mirror": f"https://www.kaggle.com/datasets/{DATASET_HANDLE}",
        "official_expected_rows": EXPECTED_ROWS,
        "source_files": source_rows,
        "publisher_engineered_rows_used": False,
    })

    return X_raw, Y, day, sample_in_day

if RAW_CKPT.exists() and not FORCE_REBUILD:
    with np.load(RAW_CKPT, allow_pickle=False) as z:
        if not validate_version(z):
            raise RuntimeError(
                "Raw checkpoint version mismatch. Set FORCE_REBUILD=True once to rebuild."
            )
        X_raw = z["X_raw"]
        Y = z["Y"]
        day = z["day"]
        sample_in_day = z["sample_in_day"]
    print("✓ Reused canonical raw checkpoint:", RAW_CKPT)
else:
    print("Building canonical raw checkpoint...")
    X_raw, Y, day, sample_in_day = build_canonical_raw_checkpoint()
    print("✓ Saved canonical raw checkpoint:", RAW_CKPT)

print("Raw shape   :", X_raw.shape)
print("Labels shape:", Y.shape)
print("Days        :", np.unique(day))

# %%
# ======================================================================================
# CELL 3 — SECURITY-BOUNDARY INFERENCE + QA
# Checkpoint: 02_boundaries.npz
# ======================================================================================

def infer_five_segments(X_day):
    midpoint = (X_day[:, 0].astype(np.float64) + X_day[:, 2].astype(np.float64)) / 2.0
    jumps = np.abs(np.diff(midpoint))

    if len(jumps) < 5:
        raise RuntimeError("Too few observations for segmentation.")

    cut_indices = np.argpartition(jumps, -4)[-4:]
    cuts = np.sort(cut_indices + 1)

    ranked = np.sort(jumps)[::-1]
    separation_ratio = ranked[3] / max(ranked[4], 1e-12)

    if separation_ratio < 10.0:
        raise RuntimeError(
            f"Security boundaries are not clearly separated: ratio={separation_ratio:.2f}"
        )

    segment = np.ones(len(X_day), dtype=np.int8)
    for cut in cuts:
        segment[cut:] += 1

    start = np.zeros(len(X_day), dtype=bool)
    start[0] = True
    start[cuts] = True

    lengths = np.diff(np.r_[0, cuts, len(X_day)]).astype(int)

    if len(np.unique(segment)) != EXPECTED_SEGMENTS_PER_DAY:
        raise RuntimeError("Expected exactly five inferred security blocks.")

    return segment, start, cuts, separation_ratio, lengths

def build_boundary_checkpoint():
    instrument_segment = np.empty(EXPECTED_ROWS, dtype=np.int8)
    is_segment_start = np.zeros(EXPECTED_ROWS, dtype=bool)
    sequence_id = np.empty(EXPECTED_ROWS, dtype=np.int16)

    boundary_rows = []

    for d in range(1, 11):
        idx = np.flatnonzero(day == d)
        seg, starts, cuts, ratio, lengths = infer_five_segments(X_raw[idx])

        instrument_segment[idx] = seg
        is_segment_start[idx] = starts
        sequence_id[idx] = (d - 1) * 5 + (seg - 1)

        boundary_rows.append({
            "day": d,
            "cuts_1indexed_within_day": (cuts + 1).tolist(),
            "segment_lengths": lengths.tolist(),
            "separation_ratio": float(ratio),
        })

    if int(is_segment_start.sum()) != EXPECTED_SEQUENCES:
        raise RuntimeError("Expected exactly 50 boundary-separated sequences.")

    if len(np.unique(sequence_id)) != EXPECTED_SEQUENCES:
        raise RuntimeError("sequence_id construction failed.")

    atomic_savez_compressed(
        BOUNDARY_CKPT,
        checkpoint_version=np.array(CHECKPOINT_VERSION),
        instrument_segment=instrument_segment,
        is_segment_start=is_segment_start,
        sequence_id=sequence_id,
    )

    atomic_json(
        RESULTS_DIR / "boundary_detection.json",
        {
            "checkpoint_version": CHECKPOINT_VERSION,
            "interpretation": (
                "Five anonymous security blocks per day inferred from dominant L1 midpoint discontinuities. "
                "Used only to prevent temporal boundary leakage; not interpreted as ticker identity."
            ),
            "days": boundary_rows,
        },
    )

    return instrument_segment, is_segment_start, sequence_id, boundary_rows

if BOUNDARY_CKPT.exists() and not FORCE_REBUILD:
    with np.load(BOUNDARY_CKPT, allow_pickle=False) as z:
        if not validate_version(z):
            raise RuntimeError(
                "Boundary checkpoint version mismatch. Set FORCE_REBUILD=True once."
            )
        instrument_segment = z["instrument_segment"]
        is_segment_start = z["is_segment_start"]
        sequence_id = z["sequence_id"]

    boundary_meta = json.loads((RESULTS_DIR / "boundary_detection.json").read_text())
    boundary_rows = boundary_meta["days"]
    print("✓ Reused boundary checkpoint:", BOUNDARY_CKPT)
else:
    instrument_segment, is_segment_start, sequence_id, boundary_rows = build_boundary_checkpoint()
    print("✓ Saved boundary checkpoint:", BOUNDARY_CKPT)

print("\nBoundary-separated sequence count:", len(np.unique(sequence_id)))
print("NOTE: these are not claimed to be statistically independent replications.")

boundary_df = pd.DataFrame(boundary_rows)
display(boundary_df)

if (boundary_df["separation_ratio"] < 10).any():
    raise RuntimeError("Boundary-separation QA failed.")

# %%
# ======================================================================================
# CELL 4 — BOUNDARY-SAFE FEATURE ENGINEERING
# Checkpoint: 03_features.parquet + 03_features.npz
# ======================================================================================

def safe_divide(a, b):
    a = np.asarray(a, dtype=np.float64)
    b = np.asarray(b, dtype=np.float64)
    out = np.full(np.broadcast(a, b).shape, np.nan, dtype=np.float64)
    np.divide(a, b, out=out, where=np.abs(b) > 1e-15)
    return out

def previous_index_by_sequence(seq):
    prev = np.arange(len(seq), dtype=np.int64) - 1
    invalid = np.zeros(len(seq), dtype=bool)
    invalid[0] = True
    invalid[1:] = seq[1:] != seq[:-1]
    prev[invalid] = -1
    return prev

def rolling_std_by_sequence(values, seq, window=20):
    out = np.full(len(values), np.nan, dtype=np.float64)

    for sid in np.unique(seq):
        idx = np.flatnonzero(seq == sid)
        s = pd.Series(values[idx], dtype="float64")
        # Includes current return and up to previous window-1 returns.
        out[idx] = s.rolling(window=window, min_periods=window).std(ddof=0).to_numpy()

    return out

def build_features():
    ask_price = X_raw[:, 0::4].astype(np.float64, copy=False)
    ask_volume = X_raw[:, 1::4].astype(np.float64, copy=False)
    bid_price = X_raw[:, 2::4].astype(np.float64, copy=False)
    bid_volume = X_raw[:, 3::4].astype(np.float64, copy=False)

    ask1 = ask_price[:, 0]
    bid1 = bid_price[:, 0]
    askv1 = ask_volume[:, 0]
    bidv1 = bid_volume[:, 0]

    midprice = (ask1 + bid1) / 2.0
    spread = ask1 - bid1
    relative_spread = safe_divide(spread, midprice)

    ask_depth_l1 = askv1
    bid_depth_l1 = bidv1
    total_depth_l1 = ask_depth_l1 + bid_depth_l1

    ask_depth_l5 = ask_volume[:, :5].sum(axis=1)
    bid_depth_l5 = bid_volume[:, :5].sum(axis=1)
    total_depth_l5 = ask_depth_l5 + bid_depth_l5

    ask_depth_l10 = ask_volume.sum(axis=1)
    bid_depth_l10 = bid_volume.sum(axis=1)
    total_depth_l10 = ask_depth_l10 + bid_depth_l10

    # L1 imbalance.
    imbalance_l1 = safe_divide(
        bid_depth_l1 - ask_depth_l1,
        bid_depth_l1 + ask_depth_l1,
    )

    # Original nested L1-L5 imbalance kept only as an auxiliary feature.
    imbalance_l5 = safe_divide(
        bid_depth_l5 - ask_depth_l5,
        bid_depth_l5 + ask_depth_l5,
    )

    # Cleaner deeper-book primitive: levels 2-5 only, excluding L1.
    ask_depth_l2_l5 = ask_volume[:, 1:5].sum(axis=1)
    bid_depth_l2_l5 = bid_volume[:, 1:5].sum(axis=1)
    imbalance_l2_l5 = safe_divide(
        bid_depth_l2_l5 - ask_depth_l2_l5,
        bid_depth_l2_l5 + ask_depth_l2_l5,
    )

    top1_depth_share = safe_divide(total_depth_l1, total_depth_l10)

    microprice = safe_divide(
        ask1 * bidv1 + bid1 * askv1,
        askv1 + bidv1,
    )
    microprice_gap = microprice - midprice

    prev = previous_index_by_sequence(sequence_id)
    valid_prev = prev >= 0

    quote_ofi_proxy = np.full(EXPECTED_ROWS, np.nan, dtype=np.float64)
    log_mid_return_1 = np.full(EXPECTED_ROWS, np.nan, dtype=np.float64)

    i = np.flatnonzero(valid_prev)
    p = prev[i]

    bid_component = (
        (bid1[i] >= bid1[p]).astype(np.float64) * bidv1[i]
        - (bid1[i] <= bid1[p]).astype(np.float64) * bidv1[p]
    )

    ask_component = (
        (ask1[i] >= ask1[p]).astype(np.float64) * askv1[p]
        - (ask1[i] <= ask1[p]).astype(np.float64) * askv1[i]
    )

    quote_ofi_proxy[i] = bid_component + ask_component

    avg_l1_depth = 0.5 * (
        (bidv1[i] + askv1[i]) + (bidv1[p] + askv1[p])
    )

    quote_ofi_depth_norm = np.full(EXPECTED_ROWS, np.nan, dtype=np.float64)
    quote_ofi_depth_norm[i] = safe_divide(
        quote_ofi_proxy[i],
        avg_l1_depth,
    )

    log_mid_return_1[i] = np.log(midprice[i] / midprice[p])

    event_volatility_20 = rolling_std_by_sequence(
        log_mid_return_1,
        sequence_id,
        window=20,
    )

    if np.any(relative_spread <= 0):
        raise RuntimeError("Non-positive relative spread encountered.")
    if np.any(total_depth_l5 <= 0):
        raise RuntimeError("Non-positive L5 total depth encountered.")

    log_relative_spread = np.log(relative_spread)
    log_total_depth_l5 = np.log(total_depth_l5)

    data = {
        "day": day.astype(np.int8),
        "instrument_segment": instrument_segment.astype(np.int8),
        "sequence_id": sequence_id.astype(np.int16),
        "sample_in_day": sample_in_day.astype(np.int32),
        "is_segment_start": is_segment_start.astype(bool),

        "midprice": midprice,
        "spread": spread,
        "relative_spread": relative_spread,
        "log_relative_spread": log_relative_spread,

        "ask_depth_l1": ask_depth_l1,
        "bid_depth_l1": bid_depth_l1,
        "total_depth_l1": total_depth_l1,

        "ask_depth_l5": ask_depth_l5,
        "bid_depth_l5": bid_depth_l5,
        "total_depth_l5": total_depth_l5,
        "log_total_depth_l5": log_total_depth_l5,

        "ask_depth_l10": ask_depth_l10,
        "bid_depth_l10": bid_depth_l10,
        "total_depth_l10": total_depth_l10,

        "imbalance_l1": imbalance_l1,
        "imbalance_l5": imbalance_l5,
        "imbalance_l2_l5": imbalance_l2_l5,

        "top1_depth_share": top1_depth_share,
        "microprice": microprice,
        "microprice_gap": microprice_gap,

        "quote_ofi_proxy": quote_ofi_proxy,
        "quote_ofi_depth_norm": quote_ofi_depth_norm,

        "log_mid_return_1": log_mid_return_1,
        "event_volatility_20": event_volatility_20,
    }

    for j, name in enumerate(LABEL_NAMES):
        data[name] = Y[:, j].astype(np.int8)

    df = pd.DataFrame(data)

    # Warmup validity: current row has all primary variables + regime variable.
    required_now = PRIMARY_CAUSAL_VARS + REGIME_VARS
    current_valid = df[required_now].notna().all(axis=1).to_numpy()

    # Save full feature table (including warmup NaNs).
    tmp_parquet = FEATURE_PARQUET.with_suffix(".tmp.parquet")
    df.to_parquet(tmp_parquet, index=False, compression="snappy")
    tmp_parquet.replace(FEATURE_PARQUET)

    atomic_savez_compressed(
        FEATURE_NPZ,
        checkpoint_version=np.array(CHECKPOINT_VERSION),
        primary_causal_vars=np.array(PRIMARY_CAUSAL_VARS),
        regime_vars=np.array(REGIME_VARS),
        current_valid=current_valid,
    )

    return df, current_valid

if FEATURE_PARQUET.exists() and FEATURE_NPZ.exists() and not FORCE_REBUILD:
    df = pd.read_parquet(FEATURE_PARQUET)
    with np.load(FEATURE_NPZ, allow_pickle=False) as z:
        if not validate_version(z):
            raise RuntimeError(
                "Feature checkpoint version mismatch. Set FORCE_REBUILD=True once."
            )
        current_valid = z["current_valid"]
    print("✓ Reused feature checkpoint:", FEATURE_PARQUET)
else:
    df, current_valid = build_features()
    print("✓ Saved feature checkpoints.")

print("Feature dataframe:", df.shape)
print("Current-state valid:", f"{current_valid.sum():,}", "/", f"{len(df):,}")

print("\nPrimary variable preview:")
display(
    df[
        ["day", "instrument_segment", "sample_in_day"]
        + PRIMARY_CAUSAL_VARS
        + REGIME_VARS
        + ["label_h10"]
    ].loc[current_valid].head(10)
)

# %%
# ======================================================================================
# CELL 5 — DATA QUALITY / REDUNDANCY AUDIT
# ======================================================================================

print("\n[QA] Structural and numerical checks")

qa = {}

qa["rows"] = int(len(df))
qa["days"] = int(df["day"].nunique())
qa["boundary_separated_sequences"] = int(df["sequence_id"].nunique())
qa["segment_starts"] = int(df["is_segment_start"].sum())
qa["publisher_engineered_features_used"] = False
qa["primary_causal_variables"] = PRIMARY_CAUSAL_VARS
qa["regime_variables"] = REGIME_VARS

if qa["rows"] != EXPECTED_ROWS:
    raise RuntimeError("Observation-count QA failed.")
if qa["days"] != EXPECTED_DAYS:
    raise RuntimeError("Day-count QA failed.")
if qa["boundary_separated_sequences"] != EXPECTED_SEQUENCES:
    raise RuntimeError("Sequence-count QA failed.")

# Exact microprice identity diagnostic:
# (microprice-mid)/spread = imbalance_l1/2 whenever spread nonzero.
identity_mask = np.isfinite(df["microprice_gap"]) & (np.abs(df["spread"]) > 1e-15)
lhs = df.loc[identity_mask, "microprice_gap"].to_numpy() / df.loc[identity_mask, "spread"].to_numpy()
rhs = 0.5 * df.loc[identity_mask, "imbalance_l1"].to_numpy()
qa["microprice_identity_max_abs_error"] = float(np.max(np.abs(lhs - rhs)))

# Spearman redundancy audit among primary causal variables.
primary_valid_df = df.loc[current_valid, PRIMARY_CAUSAL_VARS]
spearman = primary_valid_df.corr(method="spearman")
spearman.to_csv(RESULTS_DIR / "primary_spearman.csv")

upper = np.triu(np.ones(spearman.shape, dtype=bool), k=1)
max_abs_spearman = float(
    np.abs(spearman.to_numpy()[upper]).max()
)
qa["max_abs_spearman_primary"] = max_abs_spearman

# Ensure lags/differences will never cross a sequence boundary.
seq = df["sequence_id"].to_numpy()
cross_boundary_pairs = int(np.sum(seq[1:] != seq[:-1]))
qa["observed_sequence_boundaries_between_rows"] = cross_boundary_pairs

# Label distributions.
label_rows = []
for label in LABEL_NAMES:
    counts = df[label].value_counts(normalize=True).sort_index()
    label_rows.append({
        "label": label,
        "p_up_label1": float(counts.get(1, 0.0)),
        "p_stationary_label2": float(counts.get(2, 0.0)),
        "p_down_label3": float(counts.get(3, 0.0)),
    })
label_dist = pd.DataFrame(label_rows)
label_dist.to_csv(RESULTS_DIR / "label_distributions.csv", index=False)

# Per-day boundary ratios.
boundary_df = pd.DataFrame(boundary_rows)
qa["min_boundary_separation_ratio"] = float(boundary_df["separation_ratio"].min())
qa["median_boundary_separation_ratio"] = float(boundary_df["separation_ratio"].median())

qa["terminology_note"] = (
    "The 50 day×security blocks are boundary-separated sequences, not assumed independent replications. "
    "Day is the outer inferential block for later hypothesis testing."
)
qa["ofi_note"] = (
    "quote_ofi_depth_norm is a depth-normalized quote-change OFI proxy derived from consecutive FI-2010 representations; "
    "it is not event-level order-flow imbalance."
)
qa["time_note"] = (
    "All lags/horizons are FI-2010 representation steps, not seconds and not raw exchange-event timestamps."
)
qa["volatility_note"] = (
    "event_volatility_20 is a rolling derived regime variable and is excluded from the primary causal primitive set."
)

atomic_json(QUALITY_JSON, qa)

print(json.dumps(qa, indent=2))
print("\nLabel distributions:")
display(label_dist)

print("\nPrimary-variable Spearman matrix:")
display(spearman.round(3))

# %%
# ======================================================================================
# CELL 6 — TEMPORAL REPRESENTATION + ACF
# Checkpoint: 04_temporal_primary.npz
# ======================================================================================

TAU_MAX_PRIMARY = 5
ACF_LAGS = [1, 2, 3, 5, 10]

def sequence_safe_lag_valid_mask(seq, valid_now, tau_max):
    out = np.zeros(len(seq), dtype=bool)

    for sid in np.unique(seq):
        idx = np.flatnonzero(seq == sid)
        if len(idx) <= tau_max:
            continue

        for pos in range(tau_max, len(idx)):
            window = idx[pos - tau_max : pos + 1]
            if valid_now[window].all():
                out[idx[pos]] = True

    return out

def median_sequence_acf(values, seq, lag):
    vals = []
    for sid in np.unique(seq):
        idx = np.flatnonzero(seq == sid)
        x = values[idx]
        mask = np.isfinite(x)

        # Keep contiguous semantics: only pairs where both positions are finite.
        if len(x) <= lag:
            continue

        a = x[lag:]
        b = x[:-lag]
        m = np.isfinite(a) & np.isfinite(b)

        if m.sum() < 20:
            continue
        if np.std(a[m]) < 1e-15 or np.std(b[m]) < 1e-15:
            continue

        vals.append(np.corrcoef(a[m], b[m])[0, 1])

    return float(np.median(vals)) if vals else np.nan

if TEMPORAL_NPZ.exists() and ACF_CSV.exists() and not FORCE_REBUILD:
    with np.load(TEMPORAL_NPZ, allow_pickle=False) as z:
        if not validate_version(z):
            raise RuntimeError(
                "Temporal checkpoint version mismatch. Set FORCE_REBUILD=True once."
            )
        X_current = z["X_current"]
        X_state_with_regime = z["X_state_with_regime"]
        Y_temporal = z["Y"]
        day_temporal = z["day"]
        instrument_segment_temporal = z["instrument_segment"]
        sequence_id_temporal = z["sequence_id"]
        sample_in_day_temporal = z["sample_in_day"]
        primary_feature_names = z["feature_names"].astype(str).tolist()
        regime_feature_names = z["regime_feature_names"].astype(str).tolist()
        full_lag_valid = z["full_lag_valid"]
    acf_df = pd.read_csv(ACF_CSV)
    print("✓ Reused temporal checkpoint:", TEMPORAL_NPZ)
else:
    seq = df["sequence_id"].to_numpy()
    full_lag_valid = sequence_safe_lag_valid_mask(
        seq,
        current_valid,
        TAU_MAX_PRIMARY,
    )

    valid_idx = np.flatnonzero(current_valid)

    X_current = df.loc[current_valid, PRIMARY_CAUSAL_VARS].to_numpy(np.float64)
    X_state_with_regime = df.loc[current_valid, PRIMARY_CAUSAL_VARS + REGIME_VARS].to_numpy(np.float64)
    Y_temporal = df.loc[current_valid, LABEL_NAMES].to_numpy(np.int8)
    day_temporal = df.loc[current_valid, "day"].to_numpy(np.int8)
    instrument_segment_temporal = df.loc[current_valid, "instrument_segment"].to_numpy(np.int8)
    sequence_id_temporal = df.loc[current_valid, "sequence_id"].to_numpy(np.int16)
    sample_in_day_temporal = df.loc[current_valid, "sample_in_day"].to_numpy(np.int32)

    # Store lag-valid mask aligned to current-valid rows.
    lag_mask_current = full_lag_valid[current_valid]

    atomic_savez_compressed(
        TEMPORAL_NPZ,
        checkpoint_version=np.array(CHECKPOINT_VERSION),
        X_current=X_current,
        X_state_with_regime=X_state_with_regime,
        Y=Y_temporal,
        day=day_temporal,
        instrument_segment=instrument_segment_temporal,
        sequence_id=sequence_id_temporal,
        sample_in_day=sample_in_day_temporal,
        feature_names=np.array(PRIMARY_CAUSAL_VARS),
        regime_feature_names=np.array(REGIME_VARS),
        label_names=np.array(LABEL_NAMES),
        tau_max_primary=np.array(TAU_MAX_PRIMARY),
        full_lag_valid=lag_mask_current,
    )

    acf_rows = []
    for feature in PRIMARY_CAUSAL_VARS + REGIME_VARS:
        values = df[feature].to_numpy(np.float64)
        for lag in ACF_LAGS:
            acf_rows.append({
                "variable": feature,
                "lag": lag,
                "median_sequence_acf": median_sequence_acf(values, seq, lag),
            })

    acf_df = pd.DataFrame(acf_rows)
    acf_df.to_csv(ACF_CSV, index=False)

    print("✓ Saved temporal checkpoint:", TEMPORAL_NPZ)

print("Current-state rows:", f"{len(X_current):,}")
print("Primary variables :", X_current.shape[1])
print("State+regime vars :", X_state_with_regime.shape[1])
print("Tau max          :", TAU_MAX_PRIMARY)
print("Fully t..t-5 valid within current-state checkpoint:", f"{int(full_lag_valid.sum()):,}")

acf_pivot = acf_df.pivot(index="variable", columns="lag", values="median_sequence_acf")
display(acf_pivot.round(3))

# Hard QA: no lagged window may cross a sequence boundary.
with np.load(TEMPORAL_NPZ, allow_pickle=False) as z:
    seq_t = z["sequence_id"]
    mask_t = z["full_lag_valid"]

for i in np.flatnonzero(mask_t):
    if i < TAU_MAX_PRIMARY:
        raise RuntimeError("Lag-valid row too early in temporal checkpoint.")
    if not np.all(seq_t[i-TAU_MAX_PRIMARY:i+1] == seq_t[i]):
        raise RuntimeError("Temporal leakage: lag crosses a sequence boundary.")

print("✓ Temporal boundary-leakage QA passed.")

# %%
# ======================================================================================
# CELL 7 — M1 RESULTS PLOTS (SAVED; NO MODEL LEARNING)
# ======================================================================================

# 1) Boundary separation
fig = plt.figure(figsize=(8, 4.5))
plt.bar(boundary_df["day"].astype(str), boundary_df["separation_ratio"])
plt.axhline(10.0, linestyle="--")
plt.xlabel("FI-2010 day")
plt.ylabel("4th / 5th largest midpoint-jump ratio")
plt.title("Anonymous security-boundary separation")
plt.tight_layout()
fig.savefig(RESULTS_DIR / "M1_boundary_separation.png", dpi=180)
plt.show()
plt.close(fig)

# 2) Label distributions
plot_df = label_dist.set_index("label")[
    ["p_up_label1", "p_stationary_label2", "p_down_label3"]
]
fig = plt.figure(figsize=(9, 4.5))
plot_df.plot(kind="bar", ax=plt.gca())
plt.xlabel("Prediction horizon")
plt.ylabel("Class fraction")
plt.title("FI-2010 label distributions")
plt.xticks(rotation=0)
plt.tight_layout()
fig.savefig(RESULTS_DIR / "M1_label_distributions.png", dpi=180)
plt.show()
plt.close(fig)

# 3) ACF diagnostic
fig = plt.figure(figsize=(8, 5))
plt.imshow(acf_pivot.to_numpy(), aspect="auto")
plt.colorbar(label="Median within-sequence ACF")
plt.xticks(range(len(acf_pivot.columns)), acf_pivot.columns)
plt.yticks(range(len(acf_pivot.index)), acf_pivot.index)
plt.xlabel("Representation-step lag")
plt.title("Temporal persistence of M1 state variables")
plt.tight_layout()
fig.savefig(RESULTS_DIR / "M1_ACF.png", dpi=180)
plt.show()
plt.close(fig)

# 4) Primary redundancy audit
fig = plt.figure(figsize=(7, 6))
plt.imshow(spearman.to_numpy(), vmin=-1, vmax=1, aspect="equal")
plt.colorbar(label="Spearman correlation")
plt.xticks(range(len(spearman.columns)), spearman.columns, rotation=75, ha="right")
plt.yticks(range(len(spearman.index)), spearman.index)
plt.title("Primary causal-variable redundancy audit")
plt.tight_layout()
fig.savefig(RESULTS_DIR / "M1_primary_spearman.png", dpi=180)
plt.show()
plt.close(fig)

print("Saved M1 plots to:", RESULTS_DIR)

# %%
# ======================================================================================
# CELL 8 — CREATE M1 GITHUB-READY TEXT ARTEFACTS + RUN MANIFEST
# ======================================================================================

(GITHUB_DIR / "Report").mkdir(parents=True, exist_ok=True)
(GITHUB_DIR / "Video").mkdir(parents=True, exist_ok=True)
(GITHUB_DIR / "Code" / "src").mkdir(parents=True, exist_ok=True)
(GITHUB_DIR / "Data").mkdir(parents=True, exist_ok=True)
(GITHUB_DIR / "Results").mkdir(parents=True, exist_ok=True)

readme = f"""# CausalLOB — CSE 516 Major Project M1

## Problem
Learn a stable temporal probabilistic dependency structure for short-horizon limit-order-book price formation, then use a Dynamic Bayesian Network for probabilistic inference over future price states.

## Track and scope
- Track: Applied AI
- Domain: financial market microstructure / limit order books
- Principal dataset: FI-2010, NoAuction, Decimal Precision
- Dataset observations: {EXPECTED_ROWS:,}
- Calendar span: 10 days
- Boundary-separated day×security sequences: 50
- Causal scope: observational temporal causal candidates under explicit assumptions; not experimental causal proof

## PGM mapping
### Representation
Temporal graph / DBN over spread, depth, queue imbalance, deeper-book imbalance, quote-change OFI proxy, and return. Rolling volatility is treated as a regime/sensitivity variable rather than a primary causal primitive.

### Learning
Primary structure-learning plan: LPCMCI because the series are autocorrelated and latent confounding is plausible. Parameter learning will use regularized Bayesian / parametric CPDs.

### Inference
Posterior probabilities of future price states. M1 also commits to comparing exact Variable Elimination with approximate inference under partial observation using inference error vs runtime.

## SOTA anchor
A. Gerhardus and J. Runge, **High-Recall Causal Discovery for Autocorrelated Time Series with Latent Confounders**, NeurIPS 2020, German Aerospace Center (DLR), Institute of Data Science. The paper introduces LPCMCI for causal discovery in autocorrelated time series with latent confounders.

## M1 preprocessing
The code:
1. reconstructs the canonical 10-day FI-2010 NoAuction Decimal-Precision dataset;
2. uses only raw LOB rows 1–40 and labels 145–149;
3. infers five anonymous security blocks per day only to prevent temporal leakage;
4. derives boundary-safe microstructure variables;
5. creates a cleaner causal representation with L1 imbalance separated from levels 2–5 imbalance;
6. checkpoints all expensive outputs to Google Drive.

## Important terminology
- The 50 blocks are **boundary-separated sequences**, not assumed independent replications.
- Lags/horizons are **FI-2010 representation steps**, not seconds.
- `quote_ofi_depth_norm` is a **depth-normalized quote-change OFI proxy**, not event-level OFI.

## Run
Open `Code/CausalLOB_M1_FROM_SCRATCH_COLAB.ipynb` in Google Colab and run top-to-bottom once. Later sessions reuse Drive checkpoints automatically.
"""
(GITHUB_DIR / "README.md").write_text(readme, encoding="utf-8")

data_readme = f"""# Data

## Principal dataset
FI-2010 Limit Order Book dataset, NoAuction, Decimal Precision.

Official metadata/source:
https://etsin.fairdata.fi/dataset/73eb48d7-4dbc-4a10-a52a-da745b47a649

Kaggle mirror used by the notebook:
https://www.kaggle.com/datasets/{DATASET_HANDLE}

The raw dataset is not committed to GitHub because of size. The Colab notebook downloads the data when no valid canonical checkpoint exists.

## Canonical reconstruction
- Day 1: `Train_Dst_NoAuction_DecPre_CF_1.txt`
- Days 2–10: `Test_Dst_NoAuction_DecPre_CF_1.txt` through `...CF_9.txt`
- Rows 1–40: raw 10-level LOB
- Rows 41–144: publisher engineered features (excluded from the primary pipeline)
- Rows 145–149: labels h=10,20,30,50,100
- Expected total observations: {EXPECTED_ROWS:,}

## Checkpointing
Persistent checkpoints are written to:
`/content/drive/MyDrive/CausalLOB/M1_G9/checkpoints/`
"""
(GITHUB_DIR / "Data" / "README.md").write_text(data_readme, encoding="utf-8")

code_readme = """# Code

Run `CausalLOB_M1_FROM_SCRATCH_COLAB.ipynb` in Google Colab.

The notebook is resume-safe:
- `FORCE_REBUILD=False` reuses valid Drive checkpoints.
- change to `True` only when deliberately rebuilding preprocessing.
- every checkpoint carries a version tag.

The `src/` directory is reserved for exported Python source.
"""
(GITHUB_DIR / "Code" / "README.md").write_text(code_readme, encoding="utf-8")

results_readme = """# Results

M1 contains descriptive / data-quality outputs only:
- security-boundary separation diagnostic;
- label distributions;
- temporal ACF diagnostics;
- primary-variable redundancy matrix.

No causal structure-learning or trading result is claimed at M1.
"""
(GITHUB_DIR / "Results" / "README.md").write_text(results_readme, encoding="utf-8")

video_file = GITHUB_DIR / "Video" / f"M1_{GROUP_NAME}_Video_Link.txt"
if not video_file.exists():
    video_file.write_text(
        "Paste the final video walkthrough URL here before submission.\n",
        encoding="utf-8",
    )

report_draft = f"""# M1 Proposal Draft — CausalLOB

## 1. Problem statement
Limit-order-book variables are strongly autocorrelated and mutually dependent. The project investigates which lagged market-state variables are stable antecedents of short-horizon midpoint price formation, and whether a Dynamic Bayesian Network can convert that temporal structure into calibrated probabilistic inference.

**Scope:** one domain (limit-order-book microstructure) and one principal dataset (FI-2010 NoAuction Decimal Precision).

**Assumptions:** observational temporal ordering is informative; latent confounding may remain; inferred relations are not experimental causal truth.

## 2. Representation
Primary random variables:
- log relative spread
- log total L1–L5 depth
- L1 queue imbalance
- incremental L2–L5 depth imbalance
- depth-normalized quote-change OFI proxy
- one-step log midpoint return

Regime/sensitivity variable:
- rolling 20-representation return volatility

Edges represent lagged probabilistic dependencies. The primary graph will initially use lags 1–5 representation steps.

## 3. Learning
Structure learning: LPCMCI.
Reason: FI-2010 variables are strongly autocorrelated and hidden common causes are plausible. The algorithm is designed for autocorrelated time series with latent confounders.

Parameter learning: regularized Bayesian/parametric CPDs after the graph is frozen.

## 4. Inference
Compute posterior probabilities of future price states under observed and partially observed market states.
Compare exact Variable Elimination against approximate inference using inference error versus runtime.

## 5. SOTA position
Gerhardus & Runge (NeurIPS 2020), German Aerospace Center (DLR), Institute of Data Science: *High-Recall Causal Discovery for Autocorrelated Time Series with Latent Confounders*.
It introduces LPCMCI and shows improved causal-graph recall under autocorrelation while allowing latent confounding. Our project transfers that methodology to limit-order-book state dynamics, then adds a downstream DBN inference layer.

## 6. Baselines
- Structure: AR-only temporal persistence and a simpler causal-sufficiency baseline (PCMCI+ / restricted comparison).
- Prediction/inference: marginal class prior and multinomial logistic model using the same observed state.
- Inference: exact VE vs approximate inference under masked evidence.

## 7. Initial KPIs
Representation:
- zero day/security boundary leakage;
- stability to lag window and representation choices;
- redundancy diagnostics.

Learning:
- edge stability across day blocks;
- sign stability;
- FDR-controlled support;
- synthetic-SCM graph recovery;
- incremental NLL beyond own-history AR baseline.

Inference:
- log loss, Brier score, ECE;
- inference error vs computation time under partial observation.

Secondary economic diagnostics will not replace the PGM KPIs.

## 8. Experimental plan
M1: canonical data pipeline, PGM formulation, SOTA, baselines, KPIs.
M2: temporal structure learning and falsification.
M3: DBN construction, parameter learning, exact/approximate inference.
M4: robustness, economic diagnostics, external replication if suitable data are available.

## 9. Known threats
- only 10 trading days;
- anonymous securities;
- FI-2010 representation steps are not physical time;
- OFI is a quote-change proxy;
- engineered variables can share underlying primitives;
- observational causal discovery cannot prove the true market DAG;
- day-level replication is limited.

## 10. M1 evidence
Canonical rows: {EXPECTED_ROWS:,}
Boundary-separated sequences: 50
Primary causal variables: {len(PRIMARY_CAUSAL_VARS)}
Publisher engineered features used: No
"""
(GITHUB_DIR / "Report" / f"M1_{GROUP_NAME}_Report_Draft.md").write_text(
    report_draft, encoding="utf-8"
)

# Save checkpoint/run inventory.
artifact_paths = [
    RAW_CKPT, BOUNDARY_CKPT, FEATURE_PARQUET, FEATURE_NPZ,
    TEMPORAL_NPZ, ACF_CSV, QUALITY_JSON, SOURCE_MANIFEST
]

manifest = {
    "checkpoint_version": CHECKPOINT_VERSION,
    "created_at_unix": time.time(),
    "artifacts": [],
}

for p in artifact_paths:
    manifest["artifacts"].append({
        "name": p.name,
        "exists": p.exists(),
        "size_mb": round(p.stat().st_size / 1024 / 1024, 3) if p.exists() else None,
        "path": str(p),
        "sha256": sha256_file(p) if p.exists() else None,
    })

atomic_json(RUN_MANIFEST, manifest)

print("✓ M1 GitHub text scaffold created:", GITHUB_DIR)
print("✓ Run manifest:", RUN_MANIFEST)

# %%
# ======================================================================================
# CELL 9 — FINAL M1 DATA-FOUNDATION SUMMARY
# ======================================================================================

print("=" * 100)
print("CAUSALLOB — M1 DATA FOUNDATION COMPLETE")
print("=" * 100)

summary = {
    "dataset": "FI-2010 NoAuction Decimal Precision",
    "observations": int(len(df)),
    "days": int(df["day"].nunique()),
    "boundary_separated_sequences": int(df["sequence_id"].nunique()),
    "primary_causal_variables": PRIMARY_CAUSAL_VARS,
    "regime_variables": REGIME_VARS,
    "current_state_valid_rows": int(current_valid.sum()),
    "tau_max_primary": TAU_MAX_PRIMARY,
    "publisher_engineered_features_used": False,
    "raw_checkpoint": str(RAW_CKPT),
    "feature_checkpoint": str(FEATURE_PARQUET),
    "temporal_checkpoint": str(TEMPORAL_NPZ),
    "results_dir": str(RESULTS_DIR),
    "github_scaffold": str(GITHUB_DIR),
}

print(json.dumps(summary, indent=2))

print("\nM1 methodological commitments:")
print("✓ ONE domain: limit-order-book market microstructure")
print("✓ ONE principal dataset: FI-2010 NoAuction Decimal Precision")
print("✓ Representation: boundary-safe temporal graph / later DBN")
print("✓ Learning plan: LPCMCI structure learning + regularized parameter learning")
print("✓ Inference plan: posterior price-state probabilities + exact-vs-approximate benchmark")
print("✓ Causal language restricted to observational temporal candidates")
print("✓ Day is the outer inferential block; 50 sequences are not called independent replications")
print("✓ Checkpoints prevent expensive reruns after Colab disconnects")

print(f"\nRuntime this session: {time.time() - T0:.2f} sec")
```
